```python
import pandas as pd #9

df = pd.DataFrame(
    {
        "battery": [85, 40, 62],
        "speed": [1.2, 0.8, 5.5]
    },index=["Rover-1", "Rover-2", "Drone-1"]
)
print(df)
print(df.loc["Rover-2"])
print(df.loc[["Drone-1", "Rover-1"], "battery"])
print(df.loc[df["speed"]>1.0,["battery"]])

df = df.rename(columns={'battery':'battery_pct'})
print(df)
df = df.rename(index={'Drone-1': 'UAV-1'})
print(df)
df.columns = ['battery_percent', 'speed_mps']
print(df)
```
