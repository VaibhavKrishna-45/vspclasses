```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression 
```

```python
mydata = pd.read_csv("robot_battery_data.csv")
mydata.info()
mydata.head()
mydata.tail()
```

```python
X_axis = mydata[['payload_kg', 'motor_temp_c']]
X_axis.head()
```

```python
Y_axis_axis = mydata[['battery_drain']]
Y_axis.head()
```

```python
model = LinearRegression().fit(X_axis,Y_axis)
model
```

```python
print(model.intercept_)
```

```python
model.predict([[11, 13]])
```

```python
print(model.coef_)
```

```python
sns.scatterplot(data = mydata, x = 'payload_kg', y= 'battery_drain')
plt.show()
```

```python
sns.scatterplot(data = mydata, x = 'motor_temp_c', y= 'battery_drain')
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    mydata['payload_kg'],
    mydata['motor_temp_c'],
    mydata['battery_drain']
)

ax.set_xlabel('Payload (kg)')
ax.set_ylabel('Motor Temperature (°C)')
ax.set_zlabel('Battery Drain')

plt.show()
```
